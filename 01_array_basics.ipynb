5.Reshaping and resizing

```python
arr=np.arange(1,31)
arr
```

```python
arr.reshape(6*5)

# Output: A 1D array containing 30 elements
# Because 6*5 = 30, this is equivalent to arr.reshape(30)
```

```python
arr.reshape(6,5)

# Output: A 2D array with 6 rows and 5 columns
# Total elements = 6*5 = 30
```
