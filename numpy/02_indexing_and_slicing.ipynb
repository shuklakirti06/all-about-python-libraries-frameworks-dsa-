2.NUMPY INDEXING AND SLICING


```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
slice=arr[3:,3:]
slice
```

3.BOOLEAN INDEXING

```python
arr=np.arange(1,21)
arr
```

```python
bool_index=arr%2==0
bool_index
```

```python
arr=arr[bool_index]
arr
```

```python
arr = np.arange(1, 21)

# Numbers greater than 5 and less than 15
arr[(arr > 5) & (arr < 15)]
```

```python
# Numbers less than 5 OR greater than 15
arr[(arr < 5) | (arr > 15)]
```
